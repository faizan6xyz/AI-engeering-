# Encoding vs Embedding 
    Embeddings and Encoding are both methods for converting raw data into numerical form for machine learning, but they differ in purpose and structure

    1. Encoding

        Encoding means:
            Transforming data into a machine-readable format.

        Usually:    
            rule-based
            fixed transformation
            may not contain semantic meaning

        Characteristics of Encoding
            Feature	                Encoding
            Dense?	                Often sparse
            Learned?	        Usually no
            Semantic meaning?	Weak
            Rule-based?	        Yes

        Types of Encoding
            Type	                Use
            One-hot encoding	NLP categories
            Label encoding	        Categorical values
            Positional encoding	Transformers
            Binary encoding	        Compression
            Hash encoding	        Large categorical data
            
        Use Cases of Encoding
            Application	        Example
            Categorical ML data	Gender, city
            NLP token IDs	        Word indexing
            Transformers	        Position information
            Communication systems	Data transmission

    2. Embedding

        Embedding means:
            Learning compact dense vectors that capture meaning and relationships (Converts data into dense semantic vectors )

        Characteristics of Embeddings
            Feature	                Embedding
            Dense?	                Yes
            Learned?	        Yes
            Semantic meaning?	Strong
            Trainable?	        Yes

    Core Difference
    Encoding	        Embedding
    Converts format	        Learns meaning
    Usually fixed	        Trainable
    Sparse possible	        Dense vectors
    No understanding	Semantic understanding

## ✅ Quick Summary
	Encoding = "How do we write this down numerically?" → Syntax-focused, often fixed.
	Embedding = "How do we represent meaning geometrically?" → Semantics-focused, learned.
