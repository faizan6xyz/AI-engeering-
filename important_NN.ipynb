# Important Neural Network Architectures

This notebook gives a compact guide to 10 important neural network architectures and related model families.

For each one, you will find:
- what it is
- how it works
- use cases
- advantages

## 1) Transformers

**What it is:**  
A neural network architecture based on attention that processes sequences by letting each token look at other tokens in the input.

**How it works:**  
1. Input text is tokenized.  
2. Tokens are turned into embeddings.  
3. Self-attention decides which tokens matter most to each other.  
4. Feed-forward layers refine the representation.  
5. The model predicts the next token or performs a downstream task.

**Use cases:**  
- LLMs like chatbots and code assistants  
- Translation  
- Summarization  
- Vision and multimodal systems

**Advantages:**  
- Excellent at modeling long-range relationships  
- Highly parallelizable during training  
- Works across text, images, audio, and more

## 2) State Space Models (SSM / Mamba)

**What it is:**  
A sequence model that keeps a compact internal state and updates it over time, instead of relying on full self-attention.

**How it works:**  
1. The input sequence is processed step by step.  
2. A hidden state carries information forward.  
3. The model updates that state using learned dynamics.  
4. The output depends on the current input and the stored state.

**Use cases:**  
- Long-document understanding  
- Time-series forecasting  
- Genomics  
- Efficient sequence modeling

**Advantages:**  
- Very memory efficient on long sequences  
- Fast for long-context workloads  
- Can be a strong alternative to Transformers for some tasks

## 3) Diffusion Models

**What it is:**  
Generative models that learn to create data by reversing a noise process.

**How it works:**  
1. During training, real data is gradually noised.  
2. The model learns to predict or remove that noise.  
3. During generation, the model starts from random noise.  
4. It denoises step by step until a final sample appears.

**Use cases:**  
- Text-to-image generation  
- Image editing  
- Video generation  
- 3D generation

**Advantages:**  
- Very high-quality generations  
- Stable training  
- Produces diverse outputs

## 4) Vision Transformer (ViT)

**What it is:**  
A Transformer adapted for images by splitting an image into small patches and treating each patch like a token.

**How it works:**  
1. The image is divided into fixed-size patches.  
2. Each patch becomes an embedding.  
3. Self-attention relates one patch to another.  
4. The model learns global image patterns.  
5. A classifier or decoder uses the final representation.

**Use cases:**  
- Image classification  
- Object detection  
- Medical imaging  
- Remote sensing

**Advantages:**  
- Captures global context well  
- Scales strongly with data and compute  
- Often beats CNNs on large vision datasets

## 5) Graph Neural Networks (GNNs)

**What it is:**  
A neural network family designed for graph data, where nodes are connected by edges.

**How it works:**  
1. Each node starts with features.  
2. A node collects information from its neighbors.  
3. The node updates its representation using aggregated neighbor data.  
4. Repeating this over layers spreads information through the graph.

**Use cases:**  
- Social networks  
- Recommender systems  
- Fraud detection  
- Drug discovery

**Advantages:**  
- Naturally fits relational data  
- Learns from structure, not just features  
- Useful wherever connections matter

## 6) Mixture of Experts (MoE)

**What it is:**  
A model with many expert sub-networks, where only a few experts are activated for each input.

**How it works:**  
1. A router decides which experts should handle the input.  
2. Selected experts process the token or sample.  
3. Their outputs are combined.  
4. The system gets large model capacity without always using all parameters.

**Use cases:**  
- Very large language models  
- Multitask systems  
- Efficient scaling to huge parameter counts

**Advantages:**  
- High capacity with lower active compute  
- Scales well to very large models  
- Can specialize experts for different patterns

## 7) Multimodal Networks

**What it is:**  
Models that process more than one kind of data, such as text, images, audio, or video.

**How it works:**  
1. Each modality is encoded by a suitable encoder.  
2. The encoders produce representations in a shared space.  
3. A fusion mechanism combines the information.  
4. The model generates a response or prediction.

**Use cases:**  
- Image question answering  
- Speech assistants  
- Video understanding  
- Robotics

**Advantages:**  
- Closer to human-style perception  
- More powerful for real-world tasks  
- Can combine evidence from multiple sources

## 8) Retrieval-Augmented Generation (RAG)

**What it is:**  
A system that combines retrieval with a language model so answers can be grounded in external documents.

**How it works:**  
1. A user asks a question.  
2. The system searches a knowledge base using embeddings or keyword methods.  
3. Relevant chunks are retrieved.  
4. The LLM reads the retrieved context and generates the answer.

**Use cases:**  
- Company knowledge assistants  
- PDF question answering  
- Search-powered chatbots  
- Support systems

**Advantages:**  
- Reduces hallucination  
- Uses fresh or private knowledge  
- Easier to update than retraining a model

## 9) Neural Radiance Fields (NeRFs)

**What it is:**  
A neural representation of a 3D scene that learns how light and color appear from different viewpoints.

**How it works:**  
1. The model receives 3D position and viewing direction.  
2. It predicts color and density at that point.  
3. Rays are sampled through the scene.  
4. Predictions are combined to render a 2D image.

**Use cases:**  
- 3D reconstruction  
- AR/VR  
- Digital twins  
- Scene rendering

**Advantages:**  
- Produces realistic novel views  
- Compact scene representation  
- Strong for 3D visualization tasks

## 10) Spiking Neural Networks (SNNs)

**What it is:**  
Brain-inspired neural networks where neurons communicate with discrete spikes instead of continuous activations.

**How it works:**  
1. Neurons accumulate incoming signals over time.  
2. When a threshold is reached, the neuron fires a spike.  
3. Spikes propagate to other neurons.  
4. Computation is naturally event-driven and temporal.

**Use cases:**  
- Low-power edge devices  
- Neuromorphic hardware  
- Robotics  
- Time-sensitive sensor processing

**Advantages:**  
- Very energy efficient  
- Good match for event-driven hardware  
- Biologically inspired temporal processing

## Quick Comparison

| Architecture | Main Strength | Typical Use |
|---|---|---|
| Transformers | Attention over sequences | LLMs, multimodal AI |
| SSM / Mamba | Efficient long sequences | Long text, time series |
| Diffusion | High-quality generation | Images, video |
| ViT | Global vision modeling | Image understanding |
| GNN | Relational data | Graph problems |
| MoE | Scalable capacity | Large foundation models |
| Multimodal | Multiple data types | Vision-language systems |
| RAG | Grounded answers | Search + chatbots |
| NeRF | 3D scene representation | Rendering, AR/VR |
| SNN | Low-power computation | Edge AI, neuromorphic systems |
