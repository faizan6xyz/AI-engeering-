# 🤖 AI Agents — A Deep Dive

## What is an AI Agent?
An AI agent is a system where an AI model doesn't just respond to a single question and stop — it perceives its environment, reasons about it, makes decisions, takes actions, and works toward a goal, often across multiple steps, with minimal human intervention.

Think of it this way: a regular AI (like a basic chatbot) is like asking someone a question and getting an answer. An AI agent is like hiring someone, giving them a goal, and letting them figure out how to achieve it — using tools, making plans, adjusting when things go wrong.

## The Core Components of an AI Agent

1. **The Brain (LLM / AI Model)**
At the center of every agent is a large language model (like Claude, GPT-4, Gemini, etc.). This model does the thinking — it reads inputs, reasons about what to do next, and decides which actions to take. It's not just generating text; it's acting as a decision-maker.

2. **Perception (Inputs)**
The agent takes in information from the world. This can include text, documents, images, database records, web search results, API responses, code output, and more. The richer the perception, the better the agent can understand its situation.

3. **Memory**
Agents need to remember things. There are generally four types of memory:
- **In-context memory**: what's currently in the active conversation window
- **External memory**: databases or vector stores the agent can query (like long-term memory)
- **Episodic memory**: records of past interactions and experiences
- **Semantic memory**: general knowledge baked into the model's weights

4. **Tools (Actions the agent can take)**
This is what separates agents from basic chatbots. Agents are given tools — functions they can call to interact with the world. Common tools include:
- Web search
- Code execution
- File read/write
- Sending emails or messages
- Calling APIs (weather, databases, payment systems, etc.)
- Browser control

The agent decides when and how to use these tools based on its goal.

5. **Planning & Reasoning**
Agents don't just react — they plan. Given a complex goal, the agent breaks it into sub-tasks, determines an order, and works through them. Popular reasoning frameworks include:
- **ReAct**: Reason, then Act, then observe the result, repeat
- **Chain-of-Thought**: Think step by step before acting
- **Tree of Thoughts**: Explore multiple possible solution paths

6. **Action & Execution**
After deciding what to do, the agent actually does it — calls a tool, writes code, searches the web, or delegates to another agent.

7. **Feedback Loop (Observation)**
After acting, the agent observes what happened. Did the action succeed? Did it get useful information? It then uses this observation to decide its next step. This loop — `Plan → Act → Observe → Plan again` — is the heartbeat of an agent.

## How an Agent Works: Step by Step
Imagine you give an agent this goal: `"Research the top 3 AI companies in 2025 and write a report."`

1. The agent receives the goal and plans: *"I need to search the web, gather info, compare companies, then write a report."*
2. It calls a search tool with query: `"top AI companies 2025"`
3. It observes the results — reads through them
4. It searches again for more specific details on each company
5. It synthesizes the information using its reasoning
6. It writes the report and delivers it

All of this happens autonomously, step by step, without you guiding each move.

## Types of AI Agents

### Single Agent
One AI model handling everything on its own — planning, tool use, execution, and output.

### Multi-Agent Systems
Multiple specialized agents working together. For example:
- A **Planner** agent breaks down the task
- A **Researcher** agent gathers information
- A **Writer** agent drafts the output
- A **Critic** agent reviews and improves it

Each agent is good at its specialty, and they collaborate like a team.

### Autonomous Agents
These run for long periods without human input — like an agent that monitors your inbox, processes emails, and takes actions based on rules you've set.

### Human-in-the-Loop Agents
These pause at key decision points and ask for human approval before proceeding — useful when the stakes are high or actions are irreversible.

## Key Concepts You Should Know

- **Agentic Loop**: The continuous cycle of `perceive → think → act → observe` that an agent runs through until the goal is achieved.
- **Tool Calling**: The mechanism by which the LLM triggers external functions. The model outputs a structured `"call this function with these parameters"` instruction, the system executes it, and returns the result to the model.
- **Context Window**: The agent's working memory. Everything it's currently aware of — instructions, past steps, tool results — sits here. Managing this carefully is crucial for long tasks.
- **Scaffolding**: The code/framework around the LLM that manages the loop, handles tool calls, routes messages, and keeps the agent running. Examples: LangChain, LlamaIndex, AutoGen, CrewAI.
- **Grounding**: Connecting the agent to real, current information (via search or databases) so it doesn't rely solely on its training data.

## Real-World Use Cases

- **Coding agents**: Write code, run it, read the error, fix it, repeat (e.g., Claude Code, Devin)
- **Research agents**: Browse the web, read papers, summarize findings
- **Customer support agents**: Understand queries, look up account info, resolve issues
- **Data analysis agents**: Load datasets, run analysis, generate insights and charts
- **Personal assistant agents**: Manage calendar, send emails, book appointments
- **DevOps agents**: Monitor systems, detect issues, trigger fixes

## Challenges with AI Agents

- **Reliability**: Agents can make mistakes mid-task and go off track
- **Context limits**: Long tasks can overflow the context window
- **Tool errors**: If a tool fails, the agent needs to handle it gracefully
- **Cost**: Many LLM calls in a loop can get expensive
- **Safety**: An agent with broad permissions can accidentally (or maliciously) cause real harm
- **Evaluation**: Hard to measure how well an agent is doing on open-ended tasks

# React (Reason + action )
-	The core idea is simple but powerful: make the agent explicitly write out its reasoning before taking any action, and then explicitly record what it observes after. This creates a traceable, inspectable chain of thought and action.

-	The ReAct pattern has three components that alternate:
	-	Thought — The agent reasons about the current situation out loud
	-	Action — The agent specifies what tool it will use and how
	-	Observation — The agent records what the tool returned

		### Thought ->  Action -> Observation -> Thought -> Action

# 6. Reflection
-	Reflection is the agent's ability to step back from its work and critically evaluate it. While the basic ReAct loop is reactive (responding to each observation), reflection is meta-cognitive — the agent thinks about its own thinking and performance.
	- There are two kinds of reflection:

	- In-task Reflection (during execution)
	The agent pauses mid-task to evaluate progress. This is like a surgeon pausing during an operation to confirm they're working on the right area before proceeding.
	
	- Post-task Reflection (after execution)
	After completing a task, the agent reviews what happened. This is especially useful in multi-agent systems or when the agent will tackle similar tasks in the future. 

-	This is the basis of agent learning over time. Systems like Reflexion (a research framework) store these post-task reflections in memory and use them to improve future performance — essentially giving agents the ability to learn from experience without retraining the underlying model.

# Self-Correction
- Self-correction is the agent's ability to detect errors in its own work and fix them — without a human pointing out the mistake. It's one of the most valuable and impressive capabilities an agent can have.
-	Self-correction operates at multiple levels:
	- Level 1 — Execution Error Correction : 
The most basic level. An action fails — code throws an error, an API returns a 404, a search returns nothing useful. The agent detects the failure from the feedback and tries a different approach.
	- Level 2 — Logical Error Correction :
The action succeeds technically but the result reveals a flaw in the agent's reasoning or approach. The agent catches the inconsistency.
	- Level 3 — Goal Drift Correction : 
The agent realizes it has drifted away from the original goal and corrects its trajectory.
	- Level 4 — Output Quality Correction
After producing an output, the agent reviews it critically and improves it before delivering.

- What Enables Self-Correction?
	1. Self-correction works because of three things working together:
	2. Explicit verification steps — The agent is prompted or trained to check its work, not just produce it.
	3. Rich feedback signals — Error messages, conflicting data, and failed assertions give the agent information to work with.
	4. Goal anchoring — The agent keeps the original goal in mind (via memory) so it can always check if its current output matches what was asked.

# Multi-Step Reasoning
- Multi-step reasoning is the agent's ability to solve problems that cannot be answered in a single thought — problems that require building up a chain of intermediate conclusions, each depending on the previous.
- This is fundamentally different from a simple lookup or one-shot generation. Multi-step reasoning is about intellectual scaffolding — constructing an answer piece by piece.
 ## Chain-of-Thought (CoT) Reasoning
-	The most foundational form of multi-step reasoning. The agent is prompted or trained to think through a problem step by step, writing out each intermediate conclusion before arriving at the final answer.
-	The magic is that simply asking a model to "think step by step" dramatically improves accuracy on complex problems — because it forces the model to build the reasoning scaffold rather than jumping straight to an answer.
-	Without CoT:
	1. "Q: If a train travels 60mph for 2.5 hours, how far does it go?"
	2. "A: 150 miles." ← Correct here, but fails on harder problems
-	With CoT:
	1. "Q: If a train travels 60mph for 2.5 hours, how far does it go?"
	2. "A: Let me think step by step.
	3. Distance = Speed × Time.
	4. Speed = 60 mph.
	5. Time = 2.5 hours.
	6. Distance = 60 × 2.5 = 150 miles.
	7. The answer is 150 miles." ← The intermediate steps prevent errors on complex problems

## Tree of Thoughts (ToT) Reasoning
- An advanced extension of CoT. Instead of following one single chain of reasoning, the agent explores multiple reasoning branches simultaneously and selects the most promising one.
- This is especially useful for creative tasks, planning problems, and situations where the right approach isn't obvious upfront.

## Decomposition-based Reasoning
- For very large tasks, the agent explicitly breaks the problem into sub-problems, solves each independently, and assembles the results.
- Each sub-task is handled independently, and the final step assembles all the pieces. This mirrors how humans tackle large projects


#  PROMPT ENGINEERING FOR AGENTS
- Prompt engineering for agents is fundamentally different from prompting a regular chatbot. With a chatbot, a bad prompt gives a bad answer. With an agent, a bad prompt causes cascading failures across dozens of steps, wrong tool calls, hallucinated outputs, and complete mission failure. The prompt is the agent's constitution — its law, its personality, its operating manual.

##  System Prompts
- The system prompt is the most important piece of text in any agent system. It runs invisibly before every conversation, shaping everything the agent does. Think of it as the agent's DNA — it defines identity, behavior, constraints, and capabilities.
- A well-designed system prompt for an agent has several distinct layers:

	-	Layer 1 — Identity & Role
		Tell the agent exactly who it is and what it exists to do. Be specific. Vague identity produces vague behavior.

	-	Layer 2 — Capabilities & Tools
		Tell the agent what it can do. List every tool it has access to and when to use each.

	-	Layer 3 — Behavioral Rules
		Explicit rules about how the agent should operate. These prevent the most common failure modes.

	-	Layer 4 — Output Format
		Tell the agent exactly how to structure its responses.

	-	Layer 5 — Constraints & Safety
		What the agent must never do.

## Tool Instructions
-	Tool instructions are how you tell the agent precisely what each tool does, when to use it, what inputs it expects, and what output to expect back. Poorly documented tools are one of the biggest reasons agents fail.

-	The Anatomy of a Good Tool Instruction
	- Every tool description should answer four questions:
		1. What does this tool do? — Its purpose in one clear sentence.
		2. When should I use it? — The exact conditions that warrant using this tool.
		3. What does it need? — Parameters, types, constraints, examples.
		4. What will it return? — Output format and what to do with it.

## Few-Shot Prompting
- Few-shot prompting means giving the agent worked examples of the behavior you want — showing it, not just telling it. It is one of the most reliable techniques in prompt engineering because examples are more precise than instructions.
- Zero-Shot vs Few-Shot
	- Zero-shot — You give instructions only, no examples. Works for simple tasks, often fails on nuanced ones.
	- Few-shot — You show examples of correct input/output pairs first, then give the real task.

- Principles of Good Few-Shot Examples
	1. Diversity — Cover different cases, not just easy ones. Include edge cases.
	2. Correctness — Every example must be perfect. Bad examples teach bad behavior.
	3. Relevance — Examples should closely resemble the real tasks the agent will face.
	4. Variety of difficulty — Include both simple and moderately complex examples.

## Chain of Thought (CoT) Prompting
- Chain of Thought is the practice of prompting the agent to reason step by step before producing an answer. 
- The core insight: for complex problems, the quality of the answer is determined by the quality of the reasoning path to it. Forcing explicit intermediate steps prevents the model from jumping to wrong conclusions.

## ReAct Prompting
- ReAct (Reason + Act) prompting is a specific prompt structure that interleaves reasoning and action in an explicit, structured way. It was covered conceptually earlier — here we focus on the prompt engineering side of implementing it.

## Planning Prompts
- Planning prompts are specifically designed to make agents think through a full strategy before executing anything. They are particularly valuable for long, complex, multi-step tasks where diving in without a plan leads to wasted steps, contradictions, and failures.
 
 ## Structured Outputs
- Structured outputs are how you make agents return machine-readable, predictable data instead of free-form text. This is critical when the agent's output needs to be processed by code, stored in a database, fed to another agent, or displayed in a UI.
- Gives XML output or Json output and many more 

# TOOL CALLING
- 	Tool calling is the mechanism that transforms an LLM from a text generator into an agent that can act on the world. It is the bridge between reasoning and reality.

## Function Calling — The Mechanics

- Function calling works at three levels:
	- Level 1 — Definition: You tell the API what tools exist (schemas)
	- Level 2 — Selection: The model decides which tool to call and with what parameters
	- Level 3 — Execution: Your code actually runs the tool and returns the result
- The model never directly runs code or calls APIs. It produces instructions to call a tool, and your surrounding code (the scaffolding) actually executes those instructions. The model is the brain; your code is the hands.

- Parallel Function Calling
	- Modern APIs support calling multiple tools simultaneously — when the results are independent of each other.
	- This dramatically speeds up multi-data-point tasks. The API returns multiple tool_use blocks in the same response, you run them in parallel, and return both results at once.

- Sequential Function Calling
	- When one tool call depends on the result of another, they must be sequential.
	- Good agents recognize which tool calls can be parallelized and which must be sequential — making them far more efficient.

## Tool Selection
- Tool selection is the process by which the agent decides which tool to use for a given situation. This is a reasoning problem, and the agent's quality here directly determines task success or failure.

- How the Agent Selects Tools 
	- The model's tool selection is driven by three factors:
		- Factor 1 — Tool descriptions: The richer the description, the more accurately the model matches tools to situations.
		- Factor 2 — Current context: What has the agent already done? What does it already know? What is it currently trying to accomplish?
		- Factor 3 — Training: The model has been trained on countless examples of tool use and has internalized patterns of which tools fit which situations.

- Tool Selection Failure Modes
	- Over-using search — Agent searches the web for things it already knows (stable facts, math, definitions). Wastes time and API calls.
	- Under-using tools — Agent tries to answer from memory when it should search for current data. Leads to outdated or hallucinated answers.
	- Wrong tool for the job — Agent uses code_executor for a simple web search, or web_search when it should fetch a specific known URL.
	- Tool collision — Two tools could both work. Agent picks the wrong one because descriptions overlap. Solution: make descriptions explicitly say when NOT to use each tool.

## Tool Routing
- Tool routing is the higher-level architecture decision of how tasks are directed to the right tool or the right agent. In complex systems, routing happens at multiple levels.

	- Single-Agent Routing
		- In a single agent, routing is the internal decision process — the agent reads the task and decides which tool to invoke. This is the tool selection process described above.

	- Multi-Agent Routing
		- In multi-agent systems, an orchestrator agent routes entire sub-tasks to specialized sub-agents:
		- The orchestrator doesn't do the work itself — it decomposes the task and routes each piece to the right specialist.

	- Intent-Based Routing
		- A powerful pattern: before routing, classify the user's intent, then route based on the classification.

## Tool Execution
- Tool execution is where the rubber meets the road — where the agent's decision to use a tool becomes an actual operation in the real world. This is handled by your scaffolding code, not the model itself.

- The Tool Execution Pipeline
	1. MODEL OUTPUTS tool_use block
	2. SCAFFOLDING intercepts the call
	3. VALIDATION: Are parameters valid? Is the tool available?
	4. EXECUTION: Run the actual function
	5. RESULT CAPTURE: Capture output or error
	6. FORMATTING: Format result for the model to read
	7. INJECTION: Send result back in conversation as tool_result
	8. MODEL CONTINUES with the new information

- Tool Execution — Error Handling
	- Production agents must handle tool failures gracefully. The model should receive informative error messages so it can recover
	- Informative errors let the agent self-correct. Vague errors ("something went wrong") cause the agent to get stuck.

- Tool Execution — Retry Logic
	- For unreliable tools (network calls, external APIs), build in retry logic

# Memory 
## SHORT-TERM MEMORY
- Short-term memory is everything the agent knows right now, in this session. It lives entirely inside the context window — the active working space that the LLM reads on every call.
- Short-term memory has three components:

1. Conversation History
	- Conversation history is the chronological record of everything said in the current session. Every message — user, assistant, tool calls, tool results — is appended to this list and sent to the model on every API call.
	- This is why the model seems to "remember" what you said earlier in a conversation. It doesn't actually remember — it rereads the full history every single time.
	- The entire history travels with every call. The model has no memory of its own — the history is the memory.

	- The Context Window Problem
		- Here is the fundamental tension with conversation history: Context windows are finite. Claude's context window is large (hundreds of thousands of tokens) but not infinite. As a conversation grows longer — especially with tool use — the history fills up.

	- Strategies to manage context overflow:
		1. Sliding window — Keep only the last N turns of history. Simple but loses early context.
		2. Summarization — When history gets too long, summarize older turns into a compact summary and replace them. "Earlier in the conversation: Arjun is a finance professional researching Tesla's earnings.
		3. Selective retention — Keep the first turn (usually has the goal), the last K turns (most recent context), and drop the middle.
		4. Compression — Pass the conversation to a smaller/cheaper model and ask it to compress the history into dense facts before continuing.

2. Current Task
	- The current task is the agent's active goal — what it's working on right now. While conversation history is the full record, the current task is the focused objective.
	- In simple agents, the current task is just the user's latest message. In complex agents, it's a structured object that the agent maintains and updates throughout execution.
	- Current Task in the System Prompt : One powerful pattern is to dynamically inject the current task state into the system prompt at every call, This keeps the agent anchored to its current objective even as the conversation history grows long and distracting.

3. Current Context
	- Current context is the collection of all relevant information the agent has gathered so far in the current session — not the raw conversation history, but the derived working knowledge.
	- Injecting Context into Every Prompt : At each step of the agent loop, inject the current context into the prompt so the model has full awareness . This prevents the agent from:
		1. Re-searching things it already searched
		2. Re-fetching pages it already read
		3. Forgetting conclusions it already reached
		4. Contradicting itself across steps

## Long term memory
- Long-term memory is what persists across sessions. When a user comes back tomorrow, or next week, the agent still knows them. It stores preferences, facts, past interactions, and learned knowledge in an external database — outside the context window.
- This is where agents transform from tools into persistent intelligent assistants.

1. User Preferences
- How the user likes things done. Communication style, format preferences, domain expertise level, tool preferences.
- With this stored, every future session begins with the agent already knowing how to communicate with this specific person.
2. User Facts
- Factual information about the user — who they are, what they do, their context.
3. Previous Interactions
- Records of what the agent has done for this user in the past — tasks completed, topics discussed, outputs produced.

## Memory Retrieval
- Memory retrieval is the process of pulling relevant memories from storage and injecting them into the agent's context at the right time.
- The key insight: you can't inject all memories into every prompt. A user might have thousands of stored memories. You need smart, targeted retrieval.
- Retrieval Strategies
1. Recency-based — Always load the N most recent memories. Simple, fast, but misses older relevant memories.
2. Semantic similarity — Search by meaning. Best for "what do I know that's relevant to this question?"
3. Hybrid retrieval — Combine recency AND relevance. Recent memories are always included, plus top-K semantically relevant ones.
4. Importance-weighted — Memories are scored by importance when stored. High-importance memories are always retrieved.
- Injecting Retrieved Memories into Context
	- Once retrieved, memories need to be injected into the prompt in a structured, readable way , the  block gets prepended to the system prompt at the start of every session, giving the agent immediate context about who it's working with.

## Memory Update
- Memory update is how the agent writes new information into long-term memory. This can happen in three ways:
1. Explicit Update (Agent-Triggered)
	-The agent recognizes something worth remembering and explicitly stores it using a memory tool.
2. Passive Update (Automatic Extraction)
	- At the end of each session, a dedicated extraction process reviews the conversation and extracts memories automatically — without relying on the agent to notice them in the moment.
3. Conflict Resolution (Updating Stale Memories)
	- Over time, stored memories can become outdated or contradictory. A user's job changes. They develop new preferences. Old facts become wrong.

##  Episodic Memory
- Episodic memory is memory of specific events and experiences — "what happened in this particular past interaction." It's the memory of episodes, not facts.
- How Episodic Memory is Used
	- When the agent encounters a new task, it retrieves similar past episodes and uses them to:
	1. Avoid past mistakes: "Last time I did an earnings analysis for Arjun, I forgot gross margin. Don't do that again."
	2. Replicate past successes: "For this type of report, I used this approach and got a quality score of 8. Do the same."
	3. Set accurate expectations: "This type of task took 4 minutes and 8 steps last time. Plan accordingly."
	4. Personalize based on history: "Arjun has asked about EV companies 6 times. He's clearly building a sector report. Frame this analysis in that context."

- Episodic Memory Architecture — Reflexion Pattern
	- The Reflexion framework (a research paper from 2023) formalized how episodic memory improves agents. The key insight: after every task, the agent writes a self-reflection that goes into episodic memory. Future tasks retrieve these reflections and improve from them.
	- This creates a feedback loop where agents genuinely get better over time — not by retraining the model, but by accumulating and learning from experience stored in episodic memory.

## Semantic Memory
- Semantic memory is memory of general knowledge and concepts — facts, rules, domain knowledge, learned principles. Unlike episodic memory (what happened), semantic memory is what the agent knows — its internal knowledge base.
- In humans: semantic memory is knowing that "Paris is the capital of France" — not tied to any specific experience, just a fact you know.


# Planning
## Task Decomposition
- Before an agent can act, it needs to break a complex goal into manageable pieces. Task decomposition is the process of taking a high-level objective like "research competitors and write a report" and splitting it into atomic sub-tasks: search competitor A, search competitor B, extract pricing, compare features, draft report.
- There are two main styles. 
	- Hierarchical decomposition breaks the goal top-down into a tree of sub-goals, each of which can be solved independently. 
	- Sequential decomposition chains tasks one after another, where the output of step N feeds into step N+1. 
- The key challenge is granularity — decompose too coarsely and the agent gets stuck; too finely and it wastes time on trivial steps. Good decomposition also anticipates dependencies: which steps must complete before others can begin.

### Planning Algorithms
1. ReAct (Reason + Act)
- ReAct is the most widely used pattern. The agent alternates between two operations in a tight loop:
	1. Thought → "I need to find the capital of France first."
	2. Action → calls a search tool
	3. Observation → reads the result
	4. Thought → "Now I know. Let me answer."
	5. Action → returns final response
- Each thought-action-observation cycle is one "step." The agent keeps looping until it has enough information to respond. The power of ReAct is that reasoning and acting are interleaved — the agent can adapt its plan mid-execution based on what it observes. It doesn't commit to a full plan upfront; it thinks one step at a time. This makes it flexible but also potentially slow on long tasks, since it reasons from scratch at every step.

2. Plan-and-Execute
- This pattern separates planning from execution into two distinct phases.
	- Phase 1 — Plan: A "planner" LLM call produces a full step-by-step plan upfront before any tools are called.
		- Example output: 
			1. Step 1: search X
			2. Step 2: search Y
			3. Step 3: compare results
			4. Step 4: write summary.
	- Phase 2 — Execute: An "executor" works through each step sequentially, calling tools and collecting results. The planner can be re-invoked if a step fails or new information changes the goal.
- The advantage over ReAct is efficiency — the agent knows the full roadmap before starting, so it can parallelize independent steps and avoid redundant reasoning. The downside is brittleness: if reality doesn't match the plan (a tool fails, a page is missing), the agent must re-plan from scratch.

3. Tree of Thoughts (ToT)
- Tree of Thoughts treats problem-solving as a search over a tree of possible reasoning paths. At each decision point, the agent generates multiple candidate "thoughts" (partial solutions or next steps), evaluates each one, and selects the most promising branch to continue — like a chess engine exploring move trees.
- The three key operations are generation (produce N candidate next thoughts), evaluation (score each thought — is this path leading somewhere useful?), and search (use BFS or DFS to traverse the tree).
- ToT shines on tasks where the solution space is large and wrong intermediate steps are hard to recover from — puzzles, proofs, creative writing with constraints. Its weakness is cost: generating and evaluating many branches multiplies LLM calls quickly. It's overkill for straightforward tasks.

4. Graph of Thoughts (GoT)
- Graph of Thoughts extends Tree of Thoughts by replacing the tree structure with a general directed graph. In a tree, each thought has exactly one parent. In a graph, thoughts can have multiple parents — meaning partial results from different branches can be merged before continuing.
- This enables genuinely powerful patterns: solve sub-problems in parallel, combine their outputs, then refine the merged result. For example, when sorting a large list, GoT can split it into chunks, sort each chunk independently (parallel nodes in the graph), then merge the sorted chunks — directly mirroring merge sort.
- GoT also supports loops in the graph, meaning a thought can refine itself iteratively. This makes it more expressive than ToT for tasks that involve aggregation, refinement, or parallel workstreams. The tradeoff is complexity — both in implementation and in the orchestration logic needed to manage non-linear execution.

5. Reflexion
- Reflexion adds a self-reflection layer on top of any base planning strategy. After the agent completes a task (or fails), it is prompted to reflect on what went wrong and produce a verbal critique of its own reasoning. This critique is stored in memory and injected into the context of the next attempt.
- The loop looks like this:
	1. Attempt → agent tries the task
	2. Evaluate → check if the outcome was correct (via unit tests, a grader, or heuristic)
	3. Reflect → agent writes a self-critique: "I failed because I searched the wrong keyword. Next time I should try a more specific query."
	4. Retry → next attempt begins with the reflection in context
- Reflexion turns trial-and-error into structured learning within a session — without any gradient updates or fine-tuning. It has shown strong results on coding benchmarks (HumanEval) and decision-making tasks. The limitation is that it requires multiple attempts, so it's best suited for tasks where retrying is cheap and a ground-truth signal is available for evaluation.

# Multi Agent System
- A single agent has limits — context window constraints, skill specialization gaps, and the sheer complexity of large tasks. Multi-agent systems solve this by distributing work across a network of specialized agents that communicate, coordinate, and delegate to each other.
## Agent Communication
- Agents need a shared language to exchange information. Communication can happen in several ways.
- Message passing is the most common pattern — agents send structured messages (often JSON) through a message bus, queue, or direct API call. -  - Each message typically carries a sender ID, receiver ID, task description, and payload.
- Shared memory is another approach — agents read and write to a common store (a database, vector store, or scratchpad) rather than talking directly to each other. This is looser coupling but requires careful concurrency handling.
- Blackboard architecture is a classic multi-agent pattern where agents post partial results to a shared "blackboard" and any agent capable of acting on that result picks it up. Think of it like a shared whiteboard in an office — anyone can read it, anyone can contribute.
- The format of communication matters too. Agents can exchange raw text, structured schemas, code, or even tool call specifications, depending on how tightly integrated the system is.

## Agent Coordination
- Coordination is about ensuring agents don't duplicate work, conflict with each other, or deadlock waiting for each other's results.
- Sequential coordination is the simplest — Agent A finishes, then Agent B starts. Safe but slow, since no parallelism is exploited.
- Parallel coordination runs independent agents simultaneously and aggregates results. A researcher agent and a data analyst agent can run at the same time if their tasks don't depend on each other.
- Consensus coordination is used when multiple agents must agree on an outcome — for example, three agents independently evaluate a piece of code and the result is decided by majority vote. This improves robustness and catches individual agent errors.
- Hierarchical coordination structures agents in a tree — a manager agent breaks down the goal, assigns subtasks to worker agents, collects their outputs, and synthesises the final result. This is by far the most common pattern in production systems today.

## Agent Delegation
- Delegation is the act of one agent handing off a sub-task to another agent that is better suited to handle it. The delegating agent is called the orchestrator (or manager), and the receiving agent is the sub-agent (or worker).
- Good delegation requires three things. 
	1. First, the orchestrator must know what each sub-agent is capable of — this is the agent's "skill profile" or system prompt description.
	2. Second, the delegated task must be scoped clearly enough that the sub-agent can complete it without further clarification. 
	3. Third, the orchestrator must know how to interpret and integrate the sub-agent's output back into the broader task.
- Delegation can be static (the orchestrator always routes task type X to agent Y) or dynamic (the orchestrator reasons at runtime about which agent to call, based on the task at hand). Dynamic delegation is more powerful but requires the orchestrator to have a strong model of its sub-agents' capabilities.
- A common failure mode is delegation drift — where an orchestrator delegates a task, the sub-agent returns a partial result, and the orchestrator keeps re-delegating instead of recognizing the task is fundamentally incomplete. Timeouts, retry budgets, and explicit completion criteria help prevent this.

## Agent Orchestration
- Orchestration is the top-level control layer — the system that manages the lifecycle of all agents, their inputs and outputs, execution order, error handling, and state.
- An orchestrator can itself be an LLM-based agent (dynamic orchestration) or a deterministic program (static orchestration). In practice, hybrid approaches are most reliable: a deterministic outer shell manages routing and retries, while an LLM agent handles the reasoning within each step.
- Key orchestration responsibilities include task routing (deciding which agent handles what), context management (ensuring each agent gets the right slice of information without overloading its context window), error recovery (handling failures, timeouts, and invalid outputs from sub-agents), and result aggregation (merging outputs from multiple agents into a coherent whole).
- Popular frameworks that implement orchestration include LangGraph, CrewAI, AutoGen, and LlamaIndex Workflows — each with different tradeoffs between flexibility and structure.

# MCP — Model Context Protocol
- MCP is an open standard introduced by Anthropic in late 2024 that defines how AI models connect to external tools, data sources, and services in a uniform, interoperable way. Before MCP, every tool integration was a one-off — each agent framework had its own way of defining tools, calling them, and parsing results. MCP standardizes all of this.
- The analogy often used is that MCP is to AI agents what USB-C is to devices — a universal connector so that any model can plug into any tool without custom wiring.

## MCP Client
- The MCP Client lives inside the host application (Claude, an IDE, a custom agent). It is the agent's "socket" to the outside world. Its responsibilities are to maintain persistent connections to one or more MCP servers, send requests on behalf of the AI model, receive responses, and inject results back into the model's context window.
- The client speaks JSON-RPC 2.0 — a lightweight remote procedure call protocol — over either stdio (for local servers, like a subprocess) or HTTP/SSE (for remote servers). The client doesn't know or care what a server does internally; it only cares about the protocol.
- When the host application starts, the MCP client performs an initialization handshake with each connected server — exchanging version info and capability declarations. After that, the client maintains a live connection and is ready to forward tool calls.

# MCP Server
- An MCP Server is any process that implements the MCP specification and exposes capabilities to clients. It can be a local process (running on your laptop alongside the agent), a remote service (a cloud-hosted API wrapper), or an in-process library.
- Each server declares three types of capabilities during the handshake: tools (callable functions), resources (readable data sources like files or database records), and prompts (pre-built prompt templates that the model can invoke by name).
- A server wrapping a web search API, for example, would expose one tool called search with parameters query and num_results. A server wrapping a filesystem would expose a tool called read_file, another called write_file, and resources representing actual files the model can inspect.
- The separation of server from client means you can build a tool once as an MCP server and any MCP-compatible host — Claude, GPT-4 via an MCP bridge, an open-source agent — can immediately use it.

## Tool Discovery
- When the MCP client connects to a server, the very first thing it does (after the handshake) is call tools/list. The server responds with a JSON schema describing every tool it exposes — the tool's name, description, and full parameter schema (types, required fields, defaults).
- The client caches this catalogue and surfaces it to the AI model, usually by injecting the tool schemas into the system prompt or as a structured tool list. The model can then read the descriptions and decide which tools are relevant for a given task — it doesn't need to be retrained or hardcoded with tool knowledge.
- This is the core power of tool discovery: new tools become available to the model at runtime simply by connecting a new MCP server. No code changes to the model or the host are required.

## Tool Invocation
- When the AI model decides to use a tool, invocation follows a strict four-step cycle.
	1. First, the model emits a tool call — a structured JSON object with the tool name and filled-in parameters. 
	2. Second, the MCP client intercepts this (the model never calls tools directly), validates it against the tool schema, and sends a tools/call JSON-RPC request to the appropriate server. 
	3. Third, the server executes the underlying function (runs a search, queries a database, executes code) and returns a result object. 
	4. Fourth, the MCP client injects the result back into the model's context as a tool result message, and the model continues its reasoning.
- The round-trip is synchronous from the model's perspective — it "sees" the result as if it arrived instantly, even though network latency, server computation, and protocol overhead all happened in between.
- Error handling is built into the spec: servers return structured error objects (with a code and message) so the client can surface failures clearly to the model, which can then decide to retry, try a different tool, or explain the failure to the user.
