## 1. Agent Fundamentals

### 1.1 What is an AI Agent?

An **AI Agent** is an autonomous software entity that:
- Perceives its environment through sensors/inputs
- Reasons about observations
- Acts upon that environment through outputs
- Achieves specified goals

**Key Characteristics:**
- **Autonomy**: Operates independently
- **Reactivity**: Responds to environmental changes
- **Proactivity**: Takes initiative
- **Goal-Oriented**: Works towards objectives
- **Learning**: Improves through experience

**Real-World Examples:**
- ChatGPT with tool use capabilities
- Code generation agents (GitHub Copilot)
- Research agents
- Customer service bots
- Game-playing AI (AlphaGo)

### 1.2 Agent vs LLM

**Large Language Models (LLMs):**
- Predict next token based on input
- Single request-response interaction
- Pattern matching from training data
- Context window limited memory
- Generate text without external action

**AI Agents:**
- Make decisions and take actions
- Iterative reasoning loop (observe-reason-act)
- Actively select and use tools
- Maintain state and episodic memory
- Actively pursue objectives

**Key Difference:**
LLM = Smart text generator
Agent = LLM + Memory + Tools + Planning

Agents use LLMs as reasoning engines but add:
- Tool calling mechanisms
- Memory systems
- Planning capabilities
- Error correction loops

### 1.3 Agent Lifecycle

```
INITIALIZATION → PERCEPTION → REASONING → ACTION → FEEDBACK → REPEAT/TERMINATE
```

**Phase 1: INITIALIZATION**
- Load LLM models
- Register available tools
- Initialize memory structures
- Set up environment connections

**Phase 2: PERCEPTION**
- Receive user input or observation
- Update context window
- Retrieve relevant memory
- Process environmental feedback

**Phase 3: REASONING**
- Analyze current situation
- Consider available options
- Plan next action
- Select tools to use

**Phase 4: ACTION**
- Execute selected tool/function
- Generate response
- Interact with environment
- Update state

**Phase 5: FEEDBACK**
- Observe results of action
- Compare with expectations
- Update knowledge base
- Self-correct if needed

**Phase 6: REPEAT or TERMINATE**
- If goal achieved: return final answer
- If max steps reached: terminate
- Otherwise: back to Phase 2

### 1.4 Observation → Reasoning → Action → Feedback Loop

**OBSERVATION (Perception Phase)**
```
Input: What information is available?
├─ Current state of environment
├─ User query or request
├─ Tool outputs from previous steps
└─ Historical context from memory
```

**REASONING (Thinking Phase)**
```
Process: How should I respond?
├─ Analyze the observation
├─ Consider goals and constraints
├─ Evaluate available tools
├─ Generate reasoning steps
└─ Decide on next action
```

**ACTION (Execution Phase)**
```
Do: What should I do about it?
├─ Select appropriate tool(s)
├─ Call tool with parameters
├─ Generate output/response
└─ Interact with environment
```

**FEEDBACK (Learning Phase)**
```
Learn: What happened?
├─ Observe outcome
├─ Compare with expectations
├─ Update internal state
├─ Extract lessons
└─ Adjust strategy if needed
```

### 1.5 ReAct (Reason + Act)

**Framework:** Combines explicit reasoning with tool interaction.

**Core Principle:**
Agents perform better when they externalize reasoning before acting.

**ReAct Pattern:**
```
Thought: Agent reasons through the problem
Action: Agent selects which tool to use
Action Input: Parameters for the tool
Observation: Result from tool execution
[Loop back to Thought]
...
Final Answer: Solution provided
```

**Why ReAct Works:**
1. **Interpretability**: We see each reasoning step
2. **Error Detection**: Agent can catch reasoning mistakes
3. **Better Tool Selection**: Justified tool choices
4. **Complex Tasks**: Multi-step problems solved naturally

**Example:**
```
Thought: I need Q2 2024 sales revenue. Let me query the database.
Action: query_database
Action Input: SELECT SUM(amount) FROM sales WHERE quarter='Q2' AND year=2024
Observation: $2,500,000

Thought: Should verify this includes all regions.
Action: query_database
Action Input: SELECT region, SUM(amount) FROM sales WHERE quarter='Q2' AND year=2024 GROUP BY region
Observation: North: $1M, South: $800K, West: $700K (Total: $2.5M)

Final Answer: Q2 2024 total revenue is $2,500,000
```

### 1.6 Reflection

**Definition:** Agent's ability to think about its own thinking, performance, and process.

**Types of Reflection:**

**1. Process Reflection**
- Evaluate reasoning steps taken
- Question assumptions made
- Consider alternative approaches
- Identify logical gaps

**2. Outcome Reflection**
- Is the result correct?
- Does it meet the goal?
- Is it complete?
- What could improve it?

**3. Strategy Reflection**
- Did I use the best tools?
- Should I change approach?
- What have I learned?
- How to avoid this mistake next time?

**Benefits of Reflection:**
- Identifies errors before finalizing
- Improves future decisions
- Enables continuous learning
- Builds user trust through transparency

**Example:**
```
Initial thought: The answer is 42
Reflection: Wait, let me verify this. Does this make sense given the constraints?
I made an assumption that might be wrong. Let me reconsider.
Revised answer: Actually, the correct answer is 38
```

### 1.7 Self-Correction

**Definition:** Agent's mechanism for detecting and fixing its own errors.

**Error Detection Strategies:**

**1. Fact Verification**
- Check against known facts
- Validate against constraints
- Cross-reference with multiple sources
- Confirm with external tools

**2. Internal Consistency Checks**
- Look for contradictions
- Verify logical flow
- Check if conclusions follow premises
- Validate mathematical accuracy

**3. Result Validation**
- Does output meet goal?
- Is format correct?
- Are all requirements satisfied?
- Does it pass sanity checks?

**4. Tool Result Analysis**
- Expected format returned?
- Error indicators present?
- Should I retry?
- Should I use different tool?

**Self-Correction Flow:**
```
Generate Response
    ↓
Check for Errors
    ├─ No errors → Return response ✓
    └─ Error detected → 
        ├─ Identify error type
        ├─ Generate fix
        ├─ Implement correction
        └─ Recheck → Loop back
```

### 1.8 Multi-Step Reasoning

**Definition:** Breaking complex problems into sequential, simpler reasoning steps.

**Why Multi-Step Reasoning Matters:**
- Complex problems unsolvable in one step
- Reduces hallucination by grounding each step
- Enables tool use at strategic points
- Improves accuracy of final answer
- Allows mid-way error correction

**Multi-Step Framework:**
```
Step 1: Problem Decomposition
   └─ Break goal into sub-goals

Step 2: Sub-goal Analysis
   └─ What's needed for each?

Step 3: Plan Execution
   └─ Execute steps in sequence

Step 4: Result Integration
   └─ Combine results
```

**Example: Plan 5-Day Japan Trip**

```
Step 1: Decompose
  ├─ Find flights
  ├─ Book accommodation
  ├─ Plan daily itineraries
  ├─ Check visa requirements
  └─ Create budget

Step 2: Analyze Each
  ├─ Flights: Best prices, departure times
  ├─ Hotels: Location, reviews, availability
  ├─ Activities: Top spots, duration, cost
  ├─ Visa: Required docs, timeline
  └─ Budget: Sum all costs

Step 3: Execute
  ├─ Search and book flights → $800
  ├─ Search and book hotels → $500/night × 5
  ├─ Research attractions → Create itinerary
  ├─ Check visa rules → Apply if needed
  └─ Calculate total

Step 4: Integrate
   └─ Create comprehensive trip plan
```

## 2. Prompt Engineering for Agents

### 2.1 System Prompts

**Definition:** Foundational instruction that defines agent's role, capabilities, and behavior.

**Components of Effective System Prompts:**

```
1. ROLE
   └─ Define who you are: \"You are an expert data analyst\"

2. TASK
   └─ What's your job: \"Analyze data and identify trends\"

3. CONSTRAINTS
   └─ What to avoid: \"Only use provided tools\"

4. BEHAVIOR
   └─ How to act: \"Show your reasoning. Ask for clarification.\"

5. FORMAT
   └─ Output structure: \"Respond in JSON format\"
```

**Example System Prompt:**
```
You are an expert financial advisor AI.

ROLE: Help users understand financial concepts and make informed decisions.

CAPABILITIES:
- Explain financial terms and concepts
- Analyze financial scenarios
- Provide educational resources
- Reference authoritative sources

CONSTRAINTS:
- Do NOT provide personalized investment advice
- Do NOT guarantee returns
- Always disclose uncertainties
- Focus on education, not sales
- Maintain strict confidentiality

BEHAVIOR:
- Show step-by-step reasoning
- Cite sources when referencing data
- Acknowledge limitations
- Ask clarifying questions
- Recommend professional consultation when appropriate
```

### 2.2 Tool Instructions

**Definition:** Documentation that tells agents how to use available tools effectively.

**Well-Formatted Tool Instruction:**
```json
{
  \"name\": \"search_web\",
  \"description\": \"Search the internet for current information\",
  \"parameters\": {
    \"query\": {
      \"type\": \"string\",
      \"description\": \"Search query terms\",
      \"example\": \"Python 3.12 features\"
    },
    \"num_results\": {
      \"type\": \"integer\",
      \"description\": \"Number of results to return\",
      \"min\": 1,
      \"max\": 10,
      \"default\": 5
    },
    \"source_type\": {
      \"type\": \"string\",
      \"enum\": [\"web\", \"news\", \"academic\"],
      \"default\": \"web\"
    }
  },
  \"required\": [\"query\"],
  \"usage_notes\": [
    \"Use for finding recent/current information\",
    \"Specify source_type for narrower results\",
    \"Results may take 3-5 seconds\",
    \"Returns URLs and summaries\"
  ]
}
```

**Key Elements:**
- Clear description of what tool does
- Parameter documentation (types, constraints, examples)
- Usage guidelines and best practices
- Error handling information
- Performance expectations

### 2.3 Few-Shot Prompting

**Definition:** Providing examples of desired behavior to guide the agent.

**Pattern:**
```
INSTRUCTION: [What to do]
Example 1: Input → Expected Output
Example 2: Input → Expected Output
Example 3: Input → Expected Output
Now apply to: [New Input]
```

**Example: JSON Extraction**
```
Extract product information into JSON format.

Example 1:
Input: \"iPhone 15 Pro costs $999 with 256GB storage\"
Output: {\"product\": \"iPhone 15 Pro\", \"price\": 999, \"storage\": \"256GB\"}

Example 2:
Input: \"MacBook Air M3 is $1,299 with 512GB SSD\"
Output: {\"product\": \"MacBook Air M3\", \"price\": 1299, \"storage\": \"512GB\"}

Example 3:
Input: \"iPad Pro 12.9 inch, 128GB for $999\"
Output: {\"product\": \"iPad Pro 12.9\", \"storage\": \"128GB\", \"price\": 999}

Now extract from:
Input: \"Samsung Galaxy S24 Ultra, 1TB storage, $1,399\"
```

**Benefits:**
- Improves consistency
- Reduces errors
- Guides output format
- Requires limited examples
- More effective than zero-shot

### 2.4 Chain of Thought (CoT) Prompting

**Definition:** Encouraging step-by-step reasoning explicitly.

**CoT Template:**
```
Let's think step by step.

Question: [Problem]

Step 1: [First logical step]
Step 2: [Second logical step]
Step 3: [Verify intermediate result]
...
Step N: [Conclusion]

Therefore: [Final Answer]
```

**Example: Math Problem**
```
Let's think step by step.

Question: A store has 120 items and sells 30% of them.
         How many items remain?

Step 1: Calculate 30% of 120
        30% × 120 = 0.30 × 120 = 36 items sold

Step 2: Subtract sold from total
        120 - 36 = 84 items remaining

Step 3: Verify
        36 (sold) + 84 (remaining) = 120 ✓

Therefore: 84 items remain
```

**Why CoT Works:**
- Reduces hallucination
- Improves accuracy significantly
- Makes reasoning transparent
- Enables error detection

### 2.5 ReAct Prompting

**Definition:** Structuring agent responses into explicit Thought-Action-Observation cycles.

**ReAct Prompt Template:**
```
You are a helpful AI assistant.

Available Tools: [tool1, tool2, tool3]

Use this format:
Thought: What should I do?
Action: [Select tool from available]
Action Input: [Specific parameters]
Observation: [Tool result]
(Repeat Thought/Action/Observation as needed)
Thought: I now have the answer
Final Answer: [Solution]

Question: [User's question]
Thought:
```

**Concrete Example:**
```
Thought: I need current GDP data for Japan.
Action: web_search
Action Input: \"Japan GDP 2024 nominal\"
Observation: Japan's nominal GDP in 2024 is $4.2 trillion

Thought: Good, but I should provide context about global ranking.
Action: knowledge_base_search
Action Input: \"Japan economy ranking world 2024\"
Observation: Japan has 3rd largest economy (US, China, Japan)

Thought: Now I have complete answer
Final Answer: Japan's GDP in 2024 is ~$4.2 trillion,
making it the world's 3rd largest economy
```

### 2.6 Planning Prompts

**Definition:** Helping agents create structured plans before execution.

**Planning Prompt Template:**
```
Before answering, create a plan:

1. GOAL ANALYSIS
   ├─ What is being asked?
   ├─ What's the desired output format?
   └─ Are there constraints or requirements?

2. INFORMATION GATHERING
   ├─ What information do I need?
   ├─ Which tools should I use?
   └─ In what order?

3. EXECUTION PLAN
   ├─ Step 1: [Specific action]
   ├─ Step 2: [Specific action]
   └─ Step 3: [Specific action]

4. VALIDATION STRATEGY
   ├─ How do I verify correctness?
   ├─ What could go wrong?
   └─ How do I handle errors?

Then execute the plan step by step.
```

**Benefits:**
- Reduces random or inefficient actions
- Improves systematic approach
- Enables efficient tool use
- Better handles complex tasks
- Reduces errors and rework

### 2.7 Structured Output Formats: JSON

**JSON Output Format**
```json
{
  \"response\": \"Main answer here\",
  \"confidence\": 0.95,
  \"reasoning\": \"Explanation of how I arrived at this\",
  \"sources\": [
    \"source1 URL\",
    \"source2 URL\"
  ],
  \"alternative_approaches\": [
    \"Alternative 1: ...\",
    \"Alternative 2: ...\"
  ],
  \"limitations\": \"Caveats or limitations of this answer\"
}
```

**Benefits:**
- Machine parseable
- Structured data extraction
- Easy integration with code
- Enables validation
- API-friendly

### 2.8 Structured Output Formats: XML

**XML Output Format**
```xml
<response>
  <answer>The answer here</answer>
  <confidence>0.95</confidence>
  <reasoning>
    <step1>First reasoning step</step1>
    <step2>Second reasoning step</step2>
  </reasoning>
  <sources>
    <source>source1</source>
    <source>source2</source>
  </sources>
  <metadata>
    <timestamp>2024-01-15T10:30:00Z</timestamp>
    <model>agent-v2</model>
  </metadata>
</response>
```

**Use Cases:**
- Document processing
- Complex hierarchical data
- Legacy system integration
- Schema validation

### 2.9-2.10 Function Calling & Prompts

**Function Calling Format**
```
When using tools, format as:

FUNCTION_CALL: tool_name(param1=value1, param2=value2)

Examples:
FUNCTION_CALL: search_web(query=\"Python 3.12 features\", num_results=5)
FUNCTION_CALL: calculate_sum(numbers=[1, 2, 3, 4, 5])
FUNCTION_CALL: database_query(table=\"users\", filter=\"age > 21\")
```

**Function Calling Prompt Template:**
```
You have access to these functions:

Function: calculate_average
  Parameters: numbers (list of numbers)
  Returns: Average value

Function: search_database
  Parameters: query (string), table (string)
  Returns: Matching records

When you need to use a function, call it using:
FUNCTION_CALL: function_name(param1=value1)

Example:
User: \"What's the average of 10, 20, 30?\"
FUNCTION_CALL: calculate_average(numbers=[10, 20, 30])
Result: 20
Answer: The average is 20
```
