
# AI Agents Learning Roadmap (Detailed Notes)

## 1. Agent Fundamentals
An AI Agent is an LLM-powered system that can perceive information, reason about it, take actions through tools, observe results, and continue working until a goal is achieved.

### Core Loop
Observation → Reasoning → Action → Feedback

### Key Concepts
- Agent vs LLM
- ReAct (Reason + Act)
- Reflection
- Self-correction
- Multi-step reasoning
- Goal-oriented execution

### Example Workflow
User asks: "Find the latest RTX GPU prices."
1. Understand request
2. Search web
3. Collect results
4. Compare information
5. Generate answer

---

## 2. Prompt Engineering for Agents

Prompt engineering controls agent behavior.

### Topics
- System Prompts
- User Prompts
- Few-shot Learning
- Chain of Thought
- ReAct Prompting
- Planning Prompts
- Structured Outputs

### Why Important?
Agents depend heavily on instructions to decide:
- Which tools to use
- When to stop
- How to reason

## 8. Agentic RAG

Traditional RAG:
Question → Retrieve → Generate

Agentic RAG:
Question
→ Plan
→ Retrieve
→ Verify
→ Retrieve Again
→ Generate

### Topics
- Query Rewriting
- Multi-hop Retrieval
